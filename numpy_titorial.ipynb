```python
# Multi dementional arrays

arr2D=[[4,5,6],[4,5,8],[9,10,4],[1,2,3]]
print(arr2D)
column_by_adding=np.sum(arr2D,axis=0)
print(column_by_adding)
print(np.sum(arr2D))
```

```python
# vectorization -->  Vectorization is the process of performing operations on entire NumPy arrays at once without using explicit loops.


arr=[1,2,3,4,5]
arr=np.array(arr)
arr1=arr+10  # --it use broadcasting as [1,2,3,4,5] + [10,10,10,10,10]

arr2=arr*arr
arr3=arr**2
arr4=arr3-arr2
print(arr1,arr2,arr3,arr4)
```

```python
# functions 

# Aggregation function --> single output 

arr=np.array([1,2,3,4,4])

print(np.sum(arr))
print(np.max(arr))
print(np.min(arr))
print(np.prod(arr))
print(np.argmin(arr))  #--> it will provide index of min element

print(np.argmax(arr)) 
print(np.mean(arr))
print(np.std(arr)) #standard deviation
print(np.median(arr)) 
print(np.var(arr)) # variance



```

```python
# Power function


print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr,3))
```

```python
# Log function 


print(np.log(arr)) # -->log base e

print(np.log10(arr))

print(np.exp(arr))


```

```python
# rounding functions


print(np.round(2.51))

print(np.ceil(2.7))

print(np.floor(2.51))

print(np.trunc(2.7))


print(np.unique(arr))
```
