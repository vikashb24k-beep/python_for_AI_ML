```python
## clean the data 

## handle missing values


# raw_data.csv has to clean 

df=pd.read_csv("raw_data.csv")


df

```

```python
df.isnull()  #---> show true where NAN is present otherwise false 

df.isna()   # same work as isnull()

df.isnull().sum() #  --->> count no of nulls per column

df.dropna()  # --->> it will drop the row if it contain any null values

df.dropna(axis=1) # it will drop the column if it contain any null values

df.fillna("none")  ## ---->>> it will fill the missing value with the value we write in ()


# avg_age=df['age'].mean()
# df['age'].fillna(avg_age)

cleaned_data=df.copy()

avg_age=cleaned_data['age'].mean()

cleaned_data['age']=cleaned_data['age'].fillna(avg_age)


cleaned_data





```

```python
df.ffill()  #  ---> fill the previous data to upcomming next missing data col byies

df.bfill() #---> fill the next
```
